```python
import pandas as pd
from datetime import datetime
```

```python
patients_df = pd.read_csv("patients.csv")
encounters_df = pd.read_csv("encounters.csv")
conditions_df = pd.read_csv("conditions.csv")
```

```python
# Calculate the Patient Population Size
population_size = patients_df.shape[0]
print(f"Total number of patients: {population_size}")
```

```python
# Descriptive Statistics
## a) Age Range

patients_df['BIRTHDATE'] = pd.to_datetime(patients_df['BIRTHDATE'])

patients_df['AGE'] = patients_df['BIRTHDATE'].apply(lambda x: datetime.now().year - x.year)

min_age = patients_df['AGE'].min()
max_age = patients_df['AGE'].max()

print(f"Age range: {min_age} - {max_age} years")
```

```python
## b) Gender Counts
gender_counts = patients_df['GENDER'].value_counts()
print("Gender Distribution:\n", gender_counts)
```

```python
## c) Race Counts
race_counts = patients_df['RACE'].value_counts()
print("Race Distribution:\n", race_counts)
```

```python
## d) Number of Patients Per City
city_counts = patients_df['CITY'].value_counts()

most_populated_city = city_counts.idxmax()
least_populated_city = city_counts.idxmin()

print(f"City with highest population: {most_populated_city} ({city_counts.max()} patients)")
print(f"City with lowest population: {least_populated_city} ({city_counts.min()} patients)")
```

```python
# Data Integrity Check
invalid_patient_ids = encounters_df[~encounters_df['PATIENT'].isin(patients_df['Id'])]

if invalid_patient_ids.empty:
    print("No data integrity issues found.")
else:
    print(f"Data integrity issue found: {len(invalid_patient_ids)} encounters have invalid patient IDs.")
```

```python
# Proposed Reports
## 1: Most Common Chronic Diseases
chronic_diseases = conditions_df['DESCRIPTION'].value_counts().head(10)
print("Top 10 Chronic Diseases:\n", chronic_diseases)
```

```python
## 2: Hospital Utilization (Encounters per Organization)
hospital_usage = encounters_df['ORGANIZATION'].value_counts().head(10)
print("Top 10 Most Visited Hospitals:\n", hospital_usage)
```

```python
# Display Top 10 Results in Table Format
## Chronic Disease Prevalence
chronic_diseases_df = chronic_diseases.reset_index()
chronic_diseases_df.columns = ['Disease', 'Frequency']
print(chronic_diseases_df.head(10))
```

```python
## Hospital Utilization
hospital_usage_df = hospital_usage.reset_index()
hospital_usage_df.columns = ['Hospital', 'Visit Count']
print(hospital_usage_df.head(10))
```
